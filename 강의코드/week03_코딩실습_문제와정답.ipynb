## 문제 2. 입력받아 계산하기
전체 참가자 수와 팀 수를 입력받아 팀당 인원과 남는 인원을 구하세요.

- 두 값을 `int(input(...))`으로 받습니다.
- 팀당 인원은 `//`, 남는 인원은 `%`로 계산합니다.
- `# 입력`, `# 처리`, `# 출력` 주석을 넣습니다.

`53`명을 `6`팀으로 나누면 팀당 `8`명, 남는 인원 `5`명이 나오는지 확인하세요.

### 정답

```python
# 입력
participants = int(input("전체 참가자 수: "))
teams = int(input("팀 수: "))

# 처리
people_per_team = participants // teams
remaining_people = participants % teams

# 출력
print("팀당 인원:", people_per_team, "명")
print("남는 인원:", remaining_people, "명")
```

`53`과 `6`을 입력하면 `팀당 인원: 8 명`, `남는 인원: 5 명`이 출력됩니다.

## 문제 3. 의사코드부터 시작하는 미니 프로그램
동아리 회비를 정산하는 프로그램을 만드세요.

- 먼저 **의사코드**(한국어 순서)를 세 줄 이상 적습니다.
- 회원 수와 1인당 회비를 입력받습니다.
- 전체 회비를 `회원 수 × 1인당 회비`로 계산해 출력합니다.

조건: 의사코드 포함, 변수 세 개 이상, `input()` 두 번 이상, 계산식 한 개 이상, `print()`로 결과에 단위(원)를 함께 출력.

### 정답 의사코드
1. 회원 수를 입력받아 정수로 변환한다.
2. 1인당 회비를 입력받아 정수로 변환한다.
3. 회원 수와 1인당 회비를 곱해 전체 회비를 계산한다.
4. 전체 회비를 원 단위와 함께 출력한다.

### 정답 코드

```python
# 입력
members = int(input("회원 수: "))
fee_per_member = int(input("1인당 회비: "))

# 처리
total_fee = members * fee_per_member

# 출력
print("전체 회비:", total_fee, "원")
```

## 문제 4. 나누어 풀고 합치기 — 부스 순이익 계산
한 부스의 순이익(= 전체 매출 − 전체 비용)을 구합니다. 큰 문제를 작은 문제로 나누어 각각 해결한 뒤 합치세요.

1. **문제 분해** — 세 개의 하위 문제로 나눕니다.
   - 하위 문제 ①: 음료 매출 = 음료 판매량 × 음료 단가
   - 하위 문제 ②: 간식 매출 = 간식 판매량 × 간식 단가
   - 하위 문제 ③: 전체 비용 = 재료비 + 대여비
2. **패턴 찾기** — ①과 ②에서 반복되는 계산 구조를 한 문장으로 적습니다.
3. **결합** — 하위 결과를 합쳐 `순이익 = (음료 매출 + 간식 매출) − 전체 비용`을 구합니다.
4. **의사코드 → 파이썬 구현** — 필요한 값을 `int(input(...))`으로 받아 위 순서대로 계산하고, 매출·비용·순이익을 단위와 함께 출력합니다.

확인: 음료 `120`잔×`2000`원, 간식 `80`개×`1500`원, 재료비 `150000`원, 대여비 `50000`원이면 매출 `360000`원, 비용 `200000`원, 순이익 `160000`원이 나오는지 확인하세요.

### 정답 의사코드
**패턴**: 판매량 × 단가로 매출을 구하는 계산이 음료와 간식에 똑같이 반복됩니다.

1. 음료 판매량과 음료 단가를 입력받는다.
2. 간식 판매량과 간식 단가를 입력받는다.
3. 재료비와 대여비를 입력받는다.
4. 음료 매출과 간식 매출을 각각 구한다. (하위 문제 ①, ②)
5. 두 매출을 더해 전체 매출을 구한다. (결합)
6. 재료비와 대여비를 더해 전체 비용을 구한다. (하위 문제 ③)
7. 전체 매출에서 전체 비용을 빼 순이익을 구한다. (최종 결합)
8. 전체 매출, 전체 비용, 순이익을 출력한다.

### 정답 코드

```python
# 입력
drink_qty = int(input("음료 판매량: "))
drink_price = int(input("음료 단가: "))
snack_qty = int(input("간식 판매량: "))
snack_price = int(input("간식 단가: "))
material_cost = int(input("재료비: "))
rental_cost = int(input("대여비: "))

# 처리 — 하위 문제를 각각 해결한 뒤 결합
drink_sales = drink_qty * drink_price      # 하위 문제 ①: 음료 매출
snack_sales = snack_qty * snack_price      # 하위 문제 ②: 간식 매출
total_sales = drink_sales + snack_sales    # 매출 결합
total_cost = material_cost + rental_cost   # 하위 문제 ③: 전체 비용
profit = total_sales - total_cost          # 최종 결합: 순이익

# 출력
print("전체 매출:", total_sales, "원")
print("전체 비용:", total_cost, "원")
print("순이익:", profit, "원")
```

`120`, `2000`, `80`, `1500`, `150000`, `50000`을 차례로 입력하면 `전체 매출: 360000 원`, `전체 비용: 200000 원`, `순이익: 160000 원`이 출력됩니다.
