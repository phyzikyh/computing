## 2. 참가비 계산기

### 문제 파악

| 나이 | 기본 참가비 |
|:---|:---:|
| 7세 이하 | 0원 |
| 8세 이상 19세 이하 | 8,000원 |
| 20세 이상 | 15,000원 |

- 사전 등록자는 2,000원을 깎아 줍니다. 깎은 결과가 0원보다 작으면 0원입니다.
- 나이가 0 미만이거나 120 초과이면 잘못된 입력이므로 계산하지 않습니다.
- **입력** 나이, 사전 등록 여부 / **출력** 참가비

### 분해와 알고리즘
1. 입력이 올바른지 확인한다.
2. 나이에 따라 기본 참가비를 정한다.
3. 사전 등록 할인을 적용한다.

앞 단계의 결과가 다음 단계의 입력이 되므로 **순서를 바꿀 수 없습니다.**

```text
나이가 0 미만이거나 120 초과이면 "계산 불가"를 돌려주고 끝낸다.
나이가 7 이하이면 기본 참가비는 0원, 19 이하이면 8,000원, 그 외에는 15,000원이다.
사전 등록자이면 2,000원을 빼고, 0원보다 작아지면 0원으로 한다.
최종 참가비를 돌려준다.
```

### 구현
같은 판단을 여러 사람에게 쓰므로 **함수**로 묶습니다. 실행하기 전에 세 줄의 출력을 예상해 보세요.

````python
def entry_fee(age, pre_registered):
    if age < 0 or age > 120:
        return None

    if age <= 7:
        fee = 0
    elif age <= 19:
        fee = 8000
    else:
        fee = 15000

    if pre_registered:
        fee = fee - 2000
        if fee < 0:
            fee = 0

    return fee

print(entry_fee(25, False))
print(entry_fee(8, True))
print(entry_fee(-3, False))
````

**직접 예상해 보기** 아래 세 줄의 결과를 먼저 말해 본 뒤 실행합니다. (7세 사전 등록 / 20세 사전 등록 / 130세)

````python
print(entry_fee(7, True))
print(entry_fee(20, True))
print(entry_fee(130, False))
````

**읽는 방법**
- 잘못된 나이는 `None`을 돌려주어 정상 금액과 구별합니다.
- `elif`는 위에서부터 검사하므로 7 이하를 먼저, 19 이하를 그다음에 확인해야 구간이 겹치지 않습니다.
- 할인은 기본 참가비를 정한 **뒤에** 적용해야 하므로 마지막에 둡니다.

## 3. 판매 기록 통계

### 문제 파악
- **입력** `(부스 이름, 금액)` 쌍의 리스트 / **출력** 부스별 합계, 전체 매출, 순위, 비중
- 기록이 하나도 없어도 프로그램이 멈추면 안 됩니다.

### 분해와 패턴 찾기
1. 부스별로 금액을 모은다. → 2. 큰 순서로 정리한다. → 3. 출력한다.

**패턴** 기록을 읽을 때마다 "그 부스의 합계에 금액을 더한다"는 같은 동작을 합니다.
- 부스가 처음 나오면 더할 합계가 없으므로 **0에서 시작**해야 합니다. → `get(키, 0)`

### 구현
합계를 구하는 일과 순위를 매기는 일은 서로 다른 일이므로 **함수를 둘로** 나눕니다.

````python
records = [
    ("김밥", 12000), ("떡볶이", 8000), ("음료", 3000), ("김밥", 15000),
    ("음료", 4500), ("떡볶이", 9500), ("솜사탕", 2000), ("김밥", 9000),
    ("음료", 3500), ("솜사탕", 2500),
]

def total_by_booth(records):
    totals = {}
    for booth, amount in records:
        totals[booth] = totals.get(booth, 0) + amount
    return totals

def rank_booths(totals):
    pairs = []
    for booth, money in totals.items():
        pairs.append((money, booth))
    pairs.sort(reverse=True)
    return pairs

totals = total_by_booth(records)
ranking = rank_booths(totals)
grand_total = sum(totals.values())

print("전체 매출:", f"{grand_total:,}원")
for place, pair in enumerate(ranking, start=1):
    money, booth = pair
    share = money / grand_total * 100
    print(f"{place}위 {booth} {money:,}원 ({share:.1f}%)")
````

**중간 결과 보기** 함수가 각각 무엇을 돌려주는지 확인합니다.

````python
print(totals)
print(ranking)
````

**읽는 방법**
- `(금액, 이름)` 순서의 튜플로 만든 이유는, 튜플을 정렬할 때 **첫 번째 값부터 비교**하기 때문입니다.
- `reverse=True`이므로 큰 금액이 앞에 옵니다.
- 반복문, 딕셔너리, 튜플, 정렬, 함수가 한 프로그램에서 각자 맡은 일을 합니다.

## 4. 공유 공간 나누기

### 문제 파악
가로 18m, 세로 12m인 빈 공간을 **같은 크기의 정사각형 구역**으로 남김없이 나눕니다. 한 변을 가장 길게 잡으려고 합니다.
- **입력** 가로, 세로 / **출력** 한 변의 최대 길이, 구역 수
- 가로나 세로가 0 이하이면 계산할 수 없습니다.

### 분해와 알고리즘
한 변의 길이를 정한다 → 그 길이로 구역 수를 센다.

**반복되는 질문** "이 길이로 가로와 세로가 **모두** 나누어떨어지는가?" 후보를 큰 값부터 내려가며 **처음 만족하는 값**이 최댓값입니다. → `for`와 `return`(또는 `break`)

```text
가로나 세로가 0 이하이면 "계산 불가"를 돌려준다.
가로와 세로 중 작은 값부터 1까지 후보를 하나씩 내려간다.
후보로 가로와 세로를 모두 나누어떨어지면 그 후보가 한 변의 길이이다.
구역 수는 (가로 ÷ 한 변)과 (세로 ÷ 한 변)을 곱한 값이다.
```

**후보를 따라가 보기** (18 × 12)

````python
for size in range(12, 5, -1):
    print(size, 18 % size, 12 % size)
````

- `18 % 후보`와 `12 % 후보`가 **둘 다 0**이어야 합니다. 후보 9에서는 가로만 0이어서 `and`가 필요합니다.
- 어느 후보에서 멈추는지 찾아 보세요.

````python
def largest_square(width, height):
    if width <= 0 or height <= 0:
        return None

    for size in range(min(width, height), 0, -1):
        if width % size == 0 and height % size == 0:
            return size

def zone_count(width, height):
    size = largest_square(width, height)
    if size is None:
        return None
    return (width // size) * (height // size)

print(largest_square(18, 12))
print(zone_count(18, 12))
````

**경계값으로 시험하기** 가로와 세로가 같을 때, 겹치는 약수가 1뿐일 때, 한 변이 1일 때, 잘못된 입력을 넣었습니다.

````python
cases = [
    (18, 12, 6, 6), (10, 10, 10, 1), (7, 5, 1, 35),
    (12, 1, 1, 12), (0, 5, None, None), (-4, 6, None, None),
]

for w, h, exp_size, exp_count in cases:
    size = largest_square(w, h)
    count = zone_count(w, h)
    ok = size == exp_size and count == exp_count
    print(f"{w:>3} x {h:<3} 한 변 {size!s:>4}  구역 {count!s:>3}  {'통과' if ok else '실패'}")
````

**잠깐 확인하기**

- 후보를 작은 값에서 큰 값으로 올라가며 찾으면 처음 찾은 값이 최댓값이 되지 않는 이유는 무엇입니까?
- `and`를 `or`로 바꾸면 18과 12에서 어떤 값이 나옵니까?

<details><summary>답 확인</summary>

- 1은 어떤 수도 나누어떨어지게 하므로 처음 찾은 값이 항상 1이 됩니다. 가장 작은 값을 찾게 됩니다.
- 가장 먼저 확인하는 후보 12에서 세로(12 % 12 = 0)만 만족해도 참이 되어 12가 나옵니다. 가로 18은 12로 나누어떨어지지 않으므로 잘못된 답입니다.

</details>

## 5. 개인정보 가리기

### 문제 파악
- 이름은 첫 글자와 마지막 글자만 남기고 가운데를 `*`로 바꿉니다. 두 글자 이름은 둘째 글자를 가립니다.
- 전화번호 `010-1234-5678`은 가운데 네 자리를 `****`로 바꿉니다.
- 한 글자 이름이나 형식이 틀린 번호는 가릴 수 없습니다.

### 분해와 패턴
이름 가리기와 전화번호 가리기는 서로 영향을 주지 않으므로 **함수 두 개**로 나눕니다.
**공통 패턴** 남길 부분은 인덱스와 슬라이싱으로 잘라 내고, 가릴 부분은 `*`로 채워 이어 붙입니다.

````python
def mask_name(name):
    if len(name) < 2:
        return None
    if len(name) == 2:
        return name[0] + "*"
    return name[0] + "*" * (len(name) - 2) + name[-1]

def mask_phone(phone):
    if len(phone) != 13 or phone[3] != "-" or phone[8] != "-":
        return None
    return phone[:4] + "****" + phone[8:]

applicants = [
    ("홍길동", "010-1234-5678"),
    ("김민", "010-9876-5432"),
    ("남궁민수", "010-5555-0000"),
]

for name, phone in applicants:
    print(mask_name(name), mask_phone(phone))
````

**경계값과 예외값** 두 글자 이름이 경계값이고, 한 글자·빈 문자열·하이픈 없는 번호가 예외값입니다.

````python
name_cases = [("홍길동", "홍*동"), ("김민", "김*"), ("남궁민수", "남**수"), ("김", None), ("", None)]
phone_cases = [("010-1234-5678", "010-****-5678"), ("01012345678", None), ("010-1234-567", None)]

for value, expected in name_cases:
    result = mask_name(value)
    print(f"이름 {value!r:<8} 기대 {expected!s:<6} 결과 {result!s:<6} {'통과' if result == expected else '실패'}")

for value, expected in phone_cases:
    result = mask_phone(value)
    print(f"번호 {value!r:<16} 기대 {expected!s:<14} 결과 {result!s:<14} {'통과' if result == expected else '실패'}")
````

- 전화번호에 하이픈이 없으면 `mask_phone()`은 무엇을 돌려줍니까?

<details><summary>답 확인</summary>

- 별표 개수를 `len(name) - 2`로 계산하기 때문입니다. 4 - 2 = 2입니다.
- 하이픈이 없으면 길이가 11이어서 `len(phone) != 13` 조건에 걸려 `None`을 돌려줍니다.

## 6. 테스트와 코드 검토

코드가 한 번 돌아갔다고 맞는 프로그램은 아닙니다. **기대하는 결과를 먼저 적어 두고** 실제 결과와 비교합니다. 문제 파악 단계의 규칙이 곧 테스트의 기준입니다.

### 테스트 표
참가비 규칙에서 값이 바뀌는 지점은 7과 8, 19와 20, 0과 -1, 120과 121입니다.

| 구분 | 나이 | 사전 등록 | 기대 결과 |
|:---|:---:|:---:|:---:|
| 정상값 | 25 | 아니오 | 15,000원 |
| 경계값 | 20 / 19 / 8 / 7 / 0 | 아니오 | 15,000 / 8,000 / 8,000 / 0 / 0원 |
| 정상값 | 25 | 예 | 13,000원 |
| 정상값 | 8 | 예 | 6,000원 |
| 경계값 | 5 | 예 | 0원 |
| 예외값 | -1 / 121 | 아니오 | 계산 불가 |

표를 코드로 옮겨 **한 번에** 검사합니다.

````python
cases = [
    (25, False, 15000), (20, False, 15000), (19, False, 8000),
    (8, False, 8000), (7, False, 0), (0, False, 0),
    (25, True, 13000), (8, True, 6000), (5, True, 0),
    (-1, False, None), (121, False, None),
]

for age, pre, expected in cases:
    result = entry_fee(age, pre)
    status = "통과" if result == expected else "실패"
    print(f"나이 {age:>3}  사전등록 {pre!s:<5}  기대 {expected!s:>5}  결과 {result!s:>5}  {status}")
````

**테스트가 필요한 이유** 7세 이하를 `age < 7`로 잘못 쓴 함수입니다. 나이 25로는 알 수 없습니다. 어느 값에서 틀리는지 확인합니다.

````python
def entry_fee_wrong(age):
    if age < 7:
        return 0
    elif age <= 19:
        return 8000
    else:
        return 15000

print("나이 7, 기대 0원, 결과", entry_fee_wrong(7), "원")
````
