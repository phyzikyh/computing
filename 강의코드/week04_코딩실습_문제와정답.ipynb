## 문제 1. 학생 정보 카드 만들기

학생 정보를 자료형에 맞게 저장하고 확인하는 프로그램을 작성하세요.

- 이름과 학번은 문자열, 평균 점수는 실수, 과제 제출 여부는 불리언으로 저장합니다.
- 학번은 앞자리의 `0`이 유지되도록 문자열로 저장합니다.
- 각 값과 자료형을 `print()`와 `type()`으로 출력합니다.
- 마지막에 f-string으로 이름·학번·평균 점수를 한 문장으로 출력합니다.

### 정답

```python
name = "홍길동"
student_id = "012345"        # 앞자리 0 보존 → 문자열
average_score = 88.5         # 실수
submitted = True             # 불리언

print(name, type(name))
print(student_id, type(student_id))
print(average_score, type(average_score))
print(submitted, type(submitted))

print(f"{name}({student_id}) 학생의 평균 점수는 {average_score}점입니다.")
```

실행하면 각 값과 자료형(`int`은 없고 `str`, `float`, `bool`)이 출력되고, 마지막 줄에 `김민서(012345) 학생의 평균 점수는 88.5점입니다.`가 나옵니다.

## 문제 2. 주문 금액 계산기 만들기

상품 정보를 입력받아 결제 금액을 계산하는 프로그램을 작성하세요.

- 상품명은 문자열, 가격과 수량은 정수로 입력받습니다.
- 상품 금액은 `가격 × 수량`, 최종 금액은 상품 금액에 배송비 `3000`원을 더해 계산합니다.
- 상품명·상품 금액·최종 금액을 f-string으로 출력하고, 금액에는 천 단위 쉼표를 표시합니다.

### 정답

```python
product_name = input("상품명: ")
price = int(input("상품 가격: "))
quantity = int(input("수량: "))

subtotal = price * quantity
delivery_fee = 3000
total_price = subtotal + delivery_fee

print(f"상품: {product_name}")
print(f"상품 금액: {subtotal:,}원")
print(f"최종 결제 금액: {total_price:,}원")
```

`노트북`, `5000`, `3`을 차례로 입력하면 `상품 금액: 15,000원`, `최종 결제 금액: 18,000원`이 출력됩니다.

## 문제 3. 오류가 있는 주문 계산기 고치기

주어진 코드에는 네 종류의 오류가 있습니다. 모두 고쳐 주문 금액이 바르게 나오도록 완성하세요.

| 원래 코드 | 오류 | 수정 |
|---|---|---|
| `quantity = int("수량을 입력하세요")` | `ValueError` | `int(input("수량: "))`로 실제 수를 입력받음 |
| `unit_prce * quantity` | `NameError` | 변수 이름 오타를 `unit_price`로 고침 |
| `"상품 금액: " + subtotal` | `TypeError` | 문자열+정수 대신 f-string으로 출력 |
| `f"...원)` | `SyntaxError` | 닫히지 않은 따옴표를 닫음 |

가격도 `input()`으로 받은 문자열을 `int()`로 변환해야 곱셈을 할 수 있습니다.

### 정답 코드

```python
product_name = input("상품명: ")
unit_price = int(input("상품 가격: "))   # 입력을 정수로 변환
quantity = int(input("수량: "))          # ValueError 나던 부분 → 입력받기

subtotal = unit_price * quantity          # unit_prce 오타(NameError) 수정
delivery_fee = 3000
total_price = subtotal + delivery_fee

print(f"상품: {product_name}")
print(f"상품 금액: {subtotal:,}원")       # 문자열+정수(TypeError) → f-string
print(f"배송비: {delivery_fee:,}원")
print(f"최종 결제 금액: {total_price:,}원")  # 따옴표 닫음(SyntaxError 수정)
```

`볼펜`, `1200`, `5`를 차례로 입력하면 `상품 금액: 6,000원`, `최종 결제 금액: 9,000원`이 출력됩니다.
