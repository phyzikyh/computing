## 문제 1. 영화 예매 요금 정산

나이와 상영 시간대로 예매 요금을 계산하고, 여러 예매를 한꺼번에 정산하는 프로그램을 작성하세요.

1. 코드를 쓰기 전에 처리 순서를 한국어로 세 줄 이상 적습니다.
2. 함수 `ticket_price(age, is_morning)`을 만듭니다. 12세 이하와 65세 이상은 7,000원, 그 사이 나이는 13,000원이고, 조조 상영(`is_morning`이 `True`)이면 2,000원을 깎습니다. 나이가 0 미만이거나 120을 넘으면 `None`을 돌려줍니다.
3. 나이는 `input()`으로 받아 `int()`로 바꾸고, 조조 상영 여부는 조조이면 `1`, 아니면 `0`을 입력받아 `True`나 `False`로 바꿉니다. 결과는 천 단위 쉼표를 넣어 `요금: 11,000원`처럼 출력하고, `None`이면 `잘못된 나이입니다.`를 출력합니다.
4. 예매 목록을 반복문으로 처리합니다. 잘못된 예매는 `continue`로 건너뛰고 몇 건이었는지 셉니다. 정상 예매의 요금을 모두 더해 총 결제 금액과 잘못된 예매 수를 출력합니다.
5. 표의 값으로 `(나이, 조조 여부, 기대 요금)` 튜플의 리스트를 만들고, 반복문으로 한 줄씩 `통과`나 `실패`를 출력합니다.

### 정답 의사코드

```text
1. 나이가 0 미만이거나 120을 넘으면 None을 돌려주고 끝낸다.
2. 나이가 12 이하이거나 65 이상이면 요금은 7,000원, 그 외에는 13,000원이다.
3. 조조 상영이면 요금에서 2,000원을 뺀다.
4. 요금을 돌려준다.
```

### 정답 코드

````python
def ticket_price(age, is_morning):
    if age < 0 or age > 120:
        return None
    if age <= 12 or age >= 65:
        price = 7000
    else:
        price = 13000
    if is_morning:
        price -= 2000
    return price
````

````python
age = int(input("나이: "))
is_morning = int(input("조조 상영이면 1, 아니면 0: ")) == 1

price = ticket_price(age, is_morning)
if price is None:
    print("잘못된 나이입니다.")
else:
    print(f"요금: {price:,}원")
````

`30`과 `1`을 차례로 입력하면 `요금: 11,000원`이, `-5`와 `0`을 입력하면 `잘못된 나이입니다.`가 출력됩니다.

````python
orders = [(30, False), (70, True), (8, False), (-5, False), (13, True)]

total = 0
invalid = 0
for age, is_morning in orders:
    price = ticket_price(age, is_morning)
    if price is None:
        invalid += 1
        continue
    total += price

print(f"총 결제 금액: {total:,}원")
print(f"잘못된 예매: {invalid}건")
````

총 결제 금액은 `36,000원`, 잘못된 예매는 `1건`입니다.

````python
cases = [
    (30, False, 13000), (30, True, 11000), (70, True, 5000),
    (12, False, 7000), (13, False, 13000), (64, False, 13000),
    (65, False, 7000), (-5, False, None), (121, False, None),
]

for age, is_morning, expected in cases:
    result = ticket_price(age, is_morning)
    if result == expected:
        status = "통과"
    else:
        status = "실패"
    print(f"나이 {age:>3}  조조 {is_morning!s:<5}  기대 {expected!s:>5}  결과 {result!s:>5}  {status}")
````

아홉 줄이 모두 `통과`로 출력됩니다. 12세와 13세, 64세와 65세가 요금이 바뀌는 경계값입니다.

## 문제 2. 편의점 재고 입출고 정산

재고 딕셔너리와 입출고 기록으로 하루의 재고 변화를 정산하는 프로그램을 작성하세요.

```python
stock = {"생수": 40, "라면": 25, "김밥": 12, "우유": 18}

log = [
    ("판매", "생수", 15), ("입고", "김밥", 20), ("판매", "라면", 30),
    ("판매", "우유", 5), ("판매", "과자", 3), ("입고", "라면", 10),
    ("판매", "김밥", 30), ("판매", "생수", 25),
]
```

1. 처리를 시작하기 전에 `before = stock.copy()`로 재고를 따로 보관합니다.
2. `log`의 `(종류, 상품, 수량)`을 하나씩 꺼내 처리합니다. 입고이면 재고를 늘리고, 판매이면 재고를 줄입니다.
3. 재고에 없는 상품이면 `○○: 취급하지 않는 상품`을, 판매 수량이 재고보다 많으면 `○○: 재고 부족 (재고 ○개, 판매 ○개)`를 출력하고 그 기록은 `continue`로 건너뜁니다. 건너뛴 기록의 수를 `failed`로 셉니다.
4. 상품마다 `생수: 40 → 0`처럼 처리 전과 처리 후 재고를 출력합니다.
5. 재고가 10개 미만인 상품을 `(재고, 상품명)` 튜플로 모아 재고가 적은 순서로 정렬해 출력합니다.

### 정답 코드

````python
stock = {"생수": 40, "라면": 25, "김밥": 12, "우유": 18}

log = [
    ("판매", "생수", 15), ("입고", "김밥", 20), ("판매", "라면", 30),
    ("판매", "우유", 5), ("판매", "과자", 3), ("입고", "라면", 10),
    ("판매", "김밥", 30), ("판매", "생수", 25),
]

before = stock.copy()        # 처리 전 재고를 따로 보관
failed = 0

for kind, name, qty in log:
    if name not in stock:
        print(f"{name}: 취급하지 않는 상품")
        failed += 1
        continue
    if kind == "입고":
        stock[name] += qty
    else:
        if stock[name] < qty:
            print(f"{name}: 재고 부족 (재고 {stock[name]}개, 판매 {qty}개)")
            failed += 1
            continue
        stock[name] -= qty

print("건너뛴 기록:", failed)

for name in stock:
    print(f"{name}: {before[name]} → {stock[name]}")

low = []
for name, qty in stock.items():
    if qty < 10:
        low.append((qty, name))
low.sort()
print("재고 부족 목록:", low)
````

`라면: 재고 부족 (재고 25개, 판매 30개)`와 `과자: 취급하지 않는 상품`이 출력되고 건너뛴 기록은 `2`입니다. 처리 후 재고는 생수 0, 라면 35, 김밥 2, 우유 13이며 재고 부족 목록은 `[(0, '생수'), (2, '김밥')]`입니다. 생수 판매 25개는 재고 25개와 같아 처리되므로 재고가 `0`이 됩니다. `before`는 `copy()`로 만들었기 때문에 처리 전 값이 그대로 남아 있습니다.

## 문제 3. 숫자 맞히기 전략 분석

1부터 100 사이의 숫자를 맞히는 전략을 만들어 시험하고, 직접 해 보는 게임도 만드는 프로그램을 작성하세요. 전략은 가능한 범위의 가운데 값을 추측하는 것입니다. 정답이 더 크면 범위의 아래쪽을 버리고, 더 작으면 위쪽을 버립니다.

1. 함수 `guess_game(secret)`을 만듭니다. 처음 범위는 `low = 1`, `high = 100`이고, 반복할 때마다 `mid = (low + high) // 2`를 추측합니다. 추측한 값은 리스트 `history`에 추가하고 시도 횟수 `attempts`를 1 늘립니다.
2. 추측이 정답과 같으면 반복을 끝냅니다. 정답보다 작으면 `low = mid + 1`, 크면 `high = mid - 1`로 범위를 줄입니다.
3. `return attempts, history`로 두 값을 돌려주고, 호출한 쪽에서 언패킹해 출력합니다.
4. 정답이 `1`부터 `100`까지일 때를 반복문으로 모두 시험해, 가장 많이 걸린 시도 횟수와 평균 시도 횟수를 구합니다.
5. 정답을 `58`로 정해 두고 `input()`으로 숫자를 입력받는 게임을 만듭니다. 범위를 벗어난 입력은 안내만 하고 시도 횟수에 세지 않습니다.

### 정답 코드

````python
def guess_game(secret):
    low = 1
    high = 100
    attempts = 0
    history = []

    while True:
        mid = (low + high) // 2
        attempts += 1
        history.append(mid)
        if mid == secret:
            break
        elif mid < secret:
            low = mid + 1
        else:
            high = mid - 1

    return attempts, history

for secret in [73, 50, 1, 100]:
    attempts, history = guess_game(secret)
    print(f"정답 {secret:>3}: {attempts}번  {history}")
````

정답 73은 6번 `[50, 75, 62, 68, 71, 73]`, 50은 1번 `[50]`, 1은 6번 `[50, 25, 12, 6, 3, 1]`, 100은 7번 `[50, 75, 88, 94, 97, 99, 100]`입니다.

````python
most = 0
total_attempts = 0

for secret in range(1, 101):
    attempts, history = guess_game(secret)
    total_attempts += attempts
    if attempts > most:
        most = attempts

print("가장 많이 걸린 횟수:", most)
print("평균 시도 횟수:", total_attempts / 100)
````

가장 많이 걸린 횟수는 `7`, 평균은 `5.8`입니다. 범위를 절반씩 줄이므로 100개 중에서도 최대 7번이면 찾을 수 있습니다.

````python
secret = 58
attempts = 0

while True:
    guess = int(input("숫자 입력: "))
    if guess < 1 or guess > 100:
        print("1부터 100 사이의 수를 입력하세요.")
        continue
    attempts += 1
    if guess < secret:
        print("더 큰 수입니다.")
    elif guess > secret:
        print("더 작은 수입니다.")
    else:
        print(f"정답! {attempts}번 만에 맞혔습니다.")
        break
````

`50`, `75`, `0`, `60`, `58`을 차례로 입력하면 `더 큰 수입니다.`, `더 작은 수입니다.`, `1부터 100 사이의 수를 입력하세요.`, `더 작은 수입니다.`, `정답! 4번 만에 맞혔습니다.`가 출력됩니다. `0`은 범위를 벗어났으므로 `continue`로 건너뛰어 시도 횟수에 세지 않습니다.
