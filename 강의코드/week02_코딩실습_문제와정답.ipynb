## 문제 2. 변수에 저장하고 값 바꿔 계산하기
동아리 회비를 계산합니다.

- `members`에 회원 수 `8`, `fee`에 1인당 회비 `5000`을 저장합니다.
- 전체 회비 `total`을 `members * fee`로 계산해 `print("전체 회비:", total, "원")`으로 출력합니다.
- 그 뒤 회원이 늘었다고 가정하여 `members`에 `12`를 새로 저장하고, `total`을 다시 계산해 한 번 더 출력합니다.

두 번째 출력이 첫 번째와 달라진 이유를 한 줄로 설명해 보세요.

### 정답

```python
members = 8
fee = 5000
total = members * fee
print("전체 회비:", total, "원")

members = 12
total = members * fee
print("전체 회비:", total, "원")
```

`members`가 8에서 12로 바뀐 뒤 `total`을 다시 계산했기 때문에 전체 회비가 40,000원에서 60,000원으로 달라집니다.

## 문제 3. 입력–처리–출력으로 미니 프로그램 만들기
간식을 부스에 똑같이 나누는 프로그램을 만드세요.

- 간식 `total_snacks`에 `146`, 부스 `booths`에 `8`을 저장합니다.
- 한 부스당 개수는 `//`, 남는 개수는 `%`로 계산합니다.
- `# 입력`, `# 처리`, `# 출력` 주석을 각각 한 줄씩 넣습니다.
- 결과를 `print("부스당:", ..., "개")`, `print("남는 간식:", ..., "개")` 두 줄로 출력합니다.

조건: 변수 세 개 이상, 계산식 두 개 이상, `print()` 두 번 이상, 주석 한 줄 이상을 포함합니다.

### 정답

```python
# 입력
total_snacks = 146
booths = 8

# 처리
snacks_per_booth = total_snacks // booths
remaining_snacks = total_snacks % booths

# 출력
print("부스당:", snacks_per_booth, "개")
print("남는 간식:", remaining_snacks, "개")
```

실행 결과는 `부스당: 18 개`, `남는 간식: 2 개`입니다.
