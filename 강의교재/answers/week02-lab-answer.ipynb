## 문제 4. 생수 배분 계산

전체 생수 245병을 부스 6곳에 똑같이 나눌 때 부스당 생수 수와 남는 생수 수를 `//`와 `%`로 계산하세요.

````python
total_water = 245
booth_count = 6

water_per_booth = total_water // booth_count
remaining_water = total_water % booth_count

print(f"부스당 생수: {water_per_booth}병")
print(f"남는 생수: {remaining_water}병")
````

## 문제 6. 입구별 방문객 계산

`input()`으로 예상 방문객 수와 입구 수를 입력받고 정수로 변환한 뒤 입구당 예상 방문객 수를 계산하세요. 입력 예시는 `720`, `6`입니다.

````python
visitors = int(input("예상 방문객 수: "))
entrances = int(input("입구 수: "))

visitors_per_entrance = visitors / entrances
print(f"입구당 예상 방문객: {visitors_per_entrance}")
````

## 문제 7. 남은 예산 계산

총예산 1,200,000원에서 장소 대여비 450,000원, 홍보비 230,000원, 안전용품비 170,000원을 제외한 금액을 계산하세요.

````python
total_budget = 1_200_000
venue_cost = 450_000
promotion_cost = 230_000
safety_cost = 170_000

remaining_budget = total_budget - venue_cost - promotion_cost - safety_cost
print(f"남은 예산: {remaining_budget:,}원")
````

## 문제 8. 입력·처리·출력으로 분해하기

예상 방문객 150명에 대해 방문객 50명마다 안전요원 1명이 필요할 때 필요한 안전요원 수를 `//`로 계산하세요. 입력·처리·출력 주석을 붙이세요.

````python
# 입력
expected_visitors = 150
visitors_per_staff = 50

# 처리
required_staff = expected_visitors // visitors_per_staff

# 출력
print(f"필요한 안전요원: {required_staff}명")
````

## 문제 9. 같은 패턴 재사용하기

오전 방문객 180명과 운영 시간 4시간, 오후 방문객 300명과 운영 시간 5시간을 이용해 시간당 평균 방문객 수를 각각 계산하세요.

````python
morning_visitors = 180
morning_hours = 4
afternoon_visitors = 300
afternoon_hours = 5

morning_average = morning_visitors / morning_hours
afternoon_average = afternoon_visitors / afternoon_hours

print(f"오전 1시간당 평균: {morning_average}명")
print(f"오후 1시간당 평균: {afternoon_average}명")
````

## 문제 10. 오류 수정과 미니 프로그램 완성

`NameError`와 `TypeError`를 수정하고 입력·처리·출력 주석을 추가하세요. 전체 간식 수 `125`, 부스 수 `6`을 입력하면 부스당 20개, 남는 간식 5개가 출력되어야 합니다.

````python
# 입력
total_snacks = int(input("전체 간식 수: "))
booths = int(input("부스 수: "))

# 처리
snacks_per_booth = total_snacks // booths
remaining_snacks = total_snacks % booths

# 출력
print(f"부스당 간식: {snacks_per_booth}개")
print(f"남는 간식: {remaining_snacks}개")
````
