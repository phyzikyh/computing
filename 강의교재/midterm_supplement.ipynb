# 중간고사 보충수업

슬라이드에 제시된 파이썬 코드를 직접 실행하고 결과가 나온 이유를 확인합니다. 코드 셀은 위에서 아래로 차례대로 실행합니다.

## 1. 자료형 확인

화면에 같은 `20`이 출력되어도 정수와 문자열은 서로 다른 자료형입니다. `type()`은 값의 자료형을 확인하는 함수입니다.

```python
number_value = 20
text_value = "20"

print(number_value, type(number_value))
print(text_value, type(text_value))
```

정수 `20`은 덧셈과 같은 수치 계산에 사용합니다. 문자열 `"20"`은 두 글자로 구성된 문자 정보입니다.

## 2. 변수값 변경

대입문은 오른쪽 식을 먼저 계산한 뒤 결과를 왼쪽 변수에 다시 연결합니다.

```python
points = 10
points = points + 5
print(points)
```

`points + 5`를 계산할 때 현재 값 10을 사용합니다. 계산 결과 15가 다시 `points`에 연결됩니다.

## 3. 입력값과 자료형 변환

`input()`으로 입력받은 값은 문자열입니다. 아래에서는 사용자가 `15`를 입력한 상황을 문자열로 재현합니다.

```python
age = "15"
print(age, type(age))
print(int(age) + 1)
```

`int(age)`는 문자열 `"15"`를 정수 `15`로 변환합니다. 변환한 뒤에 1을 더하면 16이 됩니다.

## 4. 몫과 나머지

`//`는 나눗셈의 몫, `%`는 나머지를 구합니다.

```python
print(17 // 5)
print(17 % 5)
```

17개를 5명에게 같은 수량으로 나누면 3개씩 배분하고 2개가 남습니다.

## 5. 복합 대입 연산자

복합 대입 연산자는 현재 변수값을 이용해 계산한 뒤 결과를 같은 변수에 저장합니다. 각 줄이 끝날 때의 값을 차례대로 확인합니다.

```python
points = 10
points += 5
print("첫 번째 변경:", points)
points *= 2
print("두 번째 변경:", points)
```

## 6. 대입과 비교

`=`는 값을 변수에 대입하고, `==`는 두 값이 같은지 비교합니다. 비교 결과는 `True` 또는 `False`입니다.

```python
age = 20
print(age == 20)
```

## 7. 논리 연산자로 조건 조합

`and`는 양쪽 조건이 모두 참일 때만 `True`가 됩니다.

```python
attendance = 85
assignment = 72
passed = attendance >= 80 and assignment >= 70
print(passed)
```

## 8. if 문의 실행 범위

조건이 거짓이면 `if` 아래에서 들여쓴 문장을 건너뜁니다. 들여쓰지 않은 문장은 조건문 밖에 있으므로 실행됩니다.

```python
height = 138

if height >= 140:
    print("탑승 가능")

print("안내 종료")
```

## 9. if-else로 두 갈래 선택하기

조건이 참이면 `if` 갈래, 거짓이면 `else` 갈래를 실행합니다. 두 갈래 가운데 반드시 하나만 실행됩니다.

```python
score = 75

if score >= 60:
    result = "합격"
else:
    result = "불합격"

print(result)
```

## 10. if-elif-else의 검사 순서

조건은 위에서 아래로 검사하며 처음 참이 된 갈래 하나만 실행합니다. 나이 조건처럼 범위가 겹치면 큰 기준부터 검사합니다.

```python
age = 22

if age >= 20:
    fee = 15000
elif age >= 14:
    fee = 12000
else:
    fee = 8000

print(fee)
```

### 잘못된 조건 순서

`age >= 14`를 먼저 검사하면 22세도 첫 번째 조건에서 참이 되어 성인 조건까지 내려가지 않습니다.

```python
age = 22

if age >= 14:
    fee = 12000
elif age >= 20:
    fee = 15000
else:
    fee = 8000

print(fee)
```

## 11. range()와 끝값

`range(start, stop)`은 시작값을 포함하고 끝값은 포함하지 않습니다.

```python
for i in range(1, 5):
    print(i)
```

## 12. for 문으로 리스트 순회하기

`for` 문은 리스트의 값을 앞에서부터 하나씩 꺼내 반복 변수에 넣습니다. 반복 변수에는 리스트 전체가 아니라 현재 처리할 원소 하나가 들어갑니다.

```python
subjects = ["국어", "영어", "수학"]

for subject in subjects:
    print(subject)
```

## 13. 리스트 순회와 누산

누산 변수는 반복문 밖에서 0으로 시작합니다. 반복할 때마다 리스트에서 꺼낸 값을 합계에 더합니다.

```python
sales = [120000, 85000, 96000]
total = 0

for money in sales:
    total += money
    print(f"money={money}, total={total}")

print("최종 합계:", total)
```

## 14. continue로 현재 반복 건너뛰기

`continue`가 실행되면 현재 반복의 남은 코드를 건너뛰고 다음 값으로 이동합니다. 반복문 전체를 종료하는 `break`와 구분해야 합니다.

```python
sales = [120000, 45000, 98000]

for money in sales:
    if money < 50000:
        continue
    print(money)
```

## 15. 함수와 반환값

`def`는 함수 정의, 괄호 안의 `age`는 매개변수, `return`은 계산 결과를 호출한 곳으로 돌려주는 역할입니다.

```python
def is_adult(age):
    return age >= 20

result = is_adult(22)
print(result)
```
